```python
import nbformat

def agregar_tag_remove_cell(nombre_archivo, tag_buscado, tag_a_agregar):
    # Cargar el notebook
    with open(nombre_archivo, 'r', encoding='utf-8') as f:
        notebook = nbformat.read(f, as_version=4)

    modificado = False  # Para verificar si se hicieron cambios

    for i, cell in enumerate(notebook.cells):
        # Asegurar que la celda tiene metadata y tags
        if 'metadata' not in cell:
            cell['metadata'] = {}
        if 'tags' not in cell.metadata:
            cell.metadata['tags'] = []

        # Mostrar los tags actuales de la celda
        print(f"Celda {i}: {cell.metadata['tags']}")

        # Si la celda tiene el tag buscado, agregar el nuevo tag si no está presente
        if tag_buscado in cell.metadata['tags']:
            print(f"Celda {i} contiene '{tag_buscado}'")
            if tag_a_agregar not in cell.metadata['tags']:
                print(f"Agregando '{tag_a_agregar}' a celda {i}")
                cell.metadata['tags'].append(tag_a_agregar)
                modificado = True

    # Guardar solo si hubo cambios
    if modificado:
        with open(nombre_archivo, 'w', encoding='utf-8') as f:
            nbformat.write(notebook, f)
        print("Se han realizado cambios y guardado el archivo.")
    else:
        print("No se encontraron celdas con el tag especificado o ya tenían el tag agregado.")

# Llamar a la función con el archivo que quieres modificar
agregar_tag_remove_cell('Module 4 Conservation of Energy.ipynb', 'solution', 'remove-cell')
```
